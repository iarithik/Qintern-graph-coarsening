### Example 1

```python
# create an instance of the VRP problem
distance_matrix = base_distance_matrix[:7, :7]
cities_num = len(distance_matrix)
vehicles_num = 3
```

```python
# obtain ground truth
vrp_gt.main({
    "distance_matrix": distance_matrix,
    "num_vehicles": vehicles_num,
    "depot": 0
})
```

```python
# solving the VRP problem with Google OR Tools (standard formulation)
reference_vrp_solver = VRP(cities_num, vehicles_num, distance_matrix, None, None)
reference_solution = reference_vrp_solver.solve()
print(reference_solution)

print("------------------------------------")
```

```python
# solving the VRP problem with Google OR Tools (Adam-Ozlem formulation)
custom_vrp_solver = VRPCustom(cities_num, vehicles_num, distance_matrix)
custom_solution = custom_vrp_solver.solve()
print(custom_solution)
```
